# Preprocessing

## check columns and values

```python
import numpy as np

print("Replacing 'انتخاب نشده' and unknown with NaN...")
train_df.replace(['انتخاب نشده', 'unknown'], np.nan, inplace=True)
print("Replacement complete.")
```

### Year built

```python
train_df['Year_built'].value_counts()
```

```python
current_year = 1404
train_df.loc[train_df['Year_built'] == 'قبل از 1370', 'Year_built'] = 1370
train_df['Year_built'] = pd.to_numeric(train_df['Year_built'], errors='coerce')

train_df['Building_Age'] = current_year - train_df['Year_built']
train_df.loc[train_df['Building_Age'] < 0, 'Building_Age'] = 0

print(train_df[['Year_built', 'Building_Age']].head())
print(train_df['Building_Age'].describe())
```

### Room

```python
train_df['Rooms'].value_counts()
```

```python
print(f"ratio of rooms more than 4 to total rooms:\n{len(train_df[train_df['Rooms'] == '+4'])/ len(train_df)*100:.2f}%")
```

```python
room_mapping = {
    'بدون اتاق': 0,
    '+4': 4  # Applying the conservative approach
}

train_df['Rooms'] = train_df['Rooms'].replace(room_mapping)
```

### Floor

```python
print("Step 1: Pre-processing the 'Floor' column...")
train_df.loc[train_df['Floor'].str.contains('انتخاب نشده', na=False), 'Floor'] = np.nan
print("Preprocessing complete.")

def process_single_floor(floor_value):
    """Processes a single floor value to extract unit_floor and total_floors."""
    if pd.isna(floor_value):
        return (np.nan, np.nan)

    value = str(floor_value).strip()

    if 'از' in value:
        parts = value.split('از')
        unit_part = parts[0].strip()
        total_part = parts[1].strip()

        # Process unit's floor
        if unit_part == 'همکف': unit_floor = 0
        elif unit_part == 'زیرهمکف': unit_floor = -1
        else: unit_floor = pd.to_numeric(unit_part, errors='coerce')

        # Process total floors
        total_floors = pd.to_numeric(total_part, errors='coerce')

        # Handle illogical swapped data like "5 از 4"
        if pd.notna(unit_floor) and pd.notna(total_floors) and unit_floor > total_floors:
            unit_floor, total_floors = total_floors, unit_floor

        return (unit_floor, total_floors)
    else:
        # Handle standalone values
        if value == 'همکف': return (0, np.nan)
        elif value == 'زیرهمکف': return (-1, np.nan)
        else:
            unit_floor = pd.to_numeric(value, errors='coerce')
            return (unit_floor, np.nan)

print("Step 2: Processing function is defined.")
print("\nStep 3: Applying function and creating new features...")

floor_data = train_df['Floor'].apply(process_single_floor)
train_df[['unit_floor', 'total_floors']] = pd.DataFrame(floor_data.tolist(), index=train_df.index)

print("New features 'unit_floor' and 'total_floors' created successfully!")
print("\n" + "="*50)
print("Final Verification:")
print(train_df[['Floor', 'unit_floor', 'total_floors']].head(10))
print("\nNull values count in new columns:")
print(train_df[['unit_floor', 'total_floors']].isnull().sum())
```
